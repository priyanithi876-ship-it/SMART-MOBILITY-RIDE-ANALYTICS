```python
# ============================================================
# TASK 2 - DATA CLEANING & PREPARATION
# ============================================================

import pandas as pd

print("=" * 60)
print("TASK 2 - DATA CLEANING & PREPARATION")
print("=" * 60)

# Load dataset first
df = pd.read_csv("food delivery analytics.csv", sep=None, engine="python")

print("\nDataset loaded successfully!")
print("Original shape:", df.shape)

# ------------------------------------------------------------
# 1. Check missing values
# ------------------------------------------------------------
print("\n1. MISSING VALUES BEFORE CLEANING")
print(df.isnull().sum())

# ------------------------------------------------------------
# 2. Fill missing numeric values
# ------------------------------------------------------------
numeric_columns = [
    "Quantity",
    "OrderValue",
    "TotalAmount",
    "DeliveryTime_Minutes",
    "CustomerRating"
]

for col in numeric_columns:
    if col in df.columns and df[col].isnull().any():
        df[col] = df[col].fillna(df[col].median())

# ------------------------------------------------------------
# 3. Fill missing categorical values
# ------------------------------------------------------------
categorical_columns = [
    "CustomerID",
    "Restaurant",
    "FoodItem",
    "City",
    "PaymentMethod"
]

for col in categorical_columns:
    if col in df.columns and df[col].isnull().any():
        df[col] = df[col].fillna(df[col].mode()[0])

# ------------------------------------------------------------
# 4. Remove duplicates
# ------------------------------------------------------------
print("\n2. DUPLICATES")
print("Before:", df.duplicated().sum())

df = df.drop_duplicates()

print("After:", df.duplicated().sum())

# ------------------------------------------------------------
# 5. Validate Quantity
# ------------------------------------------------------------
print("\n3. QUANTITY VALIDATION")
print("Negative Quantity:", (df["Quantity"] < 0).sum())
print("Zero Quantity:", (df["Quantity"] == 0).sum())

# ------------------------------------------------------------
# 6. Validate OrderValue
# ------------------------------------------------------------
print("\n4. ORDERVALUE VALIDATION")
print("Negative OrderValue:", (df["OrderValue"] < 0).sum())

# ------------------------------------------------------------
# 7. Validate TotalAmount
# ------------------------------------------------------------
print("\n5. TOTALAMOUNT VALIDATION")
print("Negative TotalAmount:", (df["TotalAmount"] < 0).sum())

# ------------------------------------------------------------
# 8. Convert OrderDate to datetime
# ------------------------------------------------------------
df["OrderDate"] = pd.to_datetime(
    df["OrderDate"],
    errors="coerce"
)

print("\n6. ORDERDATE")
print("Invalid dates:", df["OrderDate"].isnull().sum())
print("Data type:", df["OrderDate"].dtype)

# ------------------------------------------------------------
# 9. Create Year, Month and Weekday
# ------------------------------------------------------------
df["Year"] = df["OrderDate"].dt.year
df["Month"] = df["OrderDate"].dt.month
df["Weekday"] = df["OrderDate"].dt.day_name()

print("\n7. NEW COLUMNS CREATED")
print("Year")
print("Month")
print("Weekday")

# ------------------------------------------------------------
# 10. Final result
# ------------------------------------------------------------
print("\nFINAL SHAPE:")
print(df.shape)

print("\nFIRST 5 ROWS AFTER CLEANING:")
display(df.head())

print("\n" + "=" * 60)
print("TASK 2 COMPLETED SUCCESSFULLY")
print("=" * 60)
```

```python
# ============================================================
# TASK 4 - ASSOCIATION RULES LEARNING
# ============================================================

import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np

from mlxtend.frequent_patterns import apriori, association_rules

print("=" * 60)
print("TASK 4 - MARKET BASKET ANALYSIS")
print("=" * 60)

# ------------------------------------------------------------
# 1. GET DATA
# ------------------------------------------------------------

if "df" in globals() and isinstance(df, pd.DataFrame):
    data = df.copy()

else:
    # Find an existing DataFrame automatically
    dfs = [
        value for value in globals().values()
        if isinstance(value, pd.DataFrame)
    ]

    if len(dfs) == 0:
        raise ValueError("No dataset found. Run your data-loading cell first.")

    data = max(dfs, key=lambda x: len(x)).copy()

print("\nDataset shape:", data.shape)

# ------------------------------------------------------------
# 2. CLEAN COLUMN NAMES
# ------------------------------------------------------------

data.columns = data.columns.astype(str).str.strip()

print("Columns:", list(data.columns))

# ------------------------------------------------------------
# 3. FIND TRANSACTION / RIDE ID
# ------------------------------------------------------------

transaction_candidates = [
    "ride_id",
    "rideid",
    "transaction_id",
    "transactionid",
    "trip_id",
    "tripid",
    "order_id",
    "orderid",
    "invoice",
    "invoice_no",
    "invoice_number"
]

transaction_col = None

for col in data.columns:
    if col.lower().replace(" ", "_") in transaction_candidates:
        transaction_col = col
        break

# If no transaction ID exists, create one
if transaction_col is None:
    data["_transaction_id"] = range(len(data))
    transaction_col = "_transaction_id"

print("Transaction column:", transaction_col)

# ------------------------------------------------------------
# 4. FIND USEFUL CATEGORICAL COLUMNS
# ------------------------------------------------------------

categorical_columns = data.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

categorical_columns = [
    col for col in categorical_columns
    if col != transaction_col
]

# Keep only columns with useful repeated values
useful_columns = []

for col in categorical_columns:
    if data[col].nunique(dropna=True) > 1:
        useful_columns.append(col)

# Use maximum 5 columns to keep basket manageable
useful_columns = useful_columns[:5]

print("Item columns used:", useful_columns)

# ------------------------------------------------------------
# 5. CREATE TRANSACTION BASKET
# ------------------------------------------------------------

records = []

for _, row in data.iterrows():

    transaction = row[transaction_col]

    for col in useful_columns:

        value = row[col]

        if pd.notna(value):

            value = str(value).strip()

            if value != "":
                item = str(col) + "_" + value
                records.append([transaction, item])

# Remove duplicates
records = pd.DataFrame(
    records,
    columns=["Transaction", "Item"]
).drop_duplicates()

# ------------------------------------------------------------
# 6. CREATE BASKET
# ------------------------------------------------------------

if records.empty:
    raise ValueError(
        "Could not create transaction basket. "
        "Please check your dataset."
    )

basket = pd.crosstab(
    records["Transaction"],
    records["Item"]
)

# VERY IMPORTANT:
# Convert 0/1 values to Boolean to remove mlxtend warning
basket = basket.astype(bool)

print("\n" + "-" * 60)
print("1. TRANSACTION BASKET")
print("-" * 60)

print("Basket shape:", basket.shape)

display(basket.head())

# ------------------------------------------------------------
# 7. FREQUENT ITEMSETS
# ------------------------------------------------------------

print("\n" + "-" * 60)
print("2. FREQUENT ITEMSETS")
print("-" * 60)

with warnings.catch_warnings():
    warnings.simplefilter("ignore")

    frequent_itemsets = apriori(
        basket,
        min_support=0.01,
        use_colnames=True
    )

# If no results, automatically try lower support
if frequent_itemsets.empty:

    with warnings.catch_warnings():
        warnings.simplefilter("ignore")

        frequent_itemsets = apriori(
            basket,
            min_support=0.005,
            use_colnames=True
        )

if frequent_itemsets.empty:

    with warnings.catch_warnings():
        warnings.simplefilter("ignore")

        frequent_itemsets = apriori(
            basket,
            min_support=0.001,
            use_colnames=True
        )

frequent_itemsets = frequent_itemsets.sort_values(
    by="support",
    ascending=False
).reset_index(drop=True)

print("Number of frequent itemsets:", len(frequent_itemsets))

if not frequent_itemsets.empty:
    display(frequent_itemsets.head(15))
else:
    print("No frequent itemsets found.")

# ------------------------------------------------------------
# 8. ASSOCIATION RULES
# ------------------------------------------------------------

print("\n" + "-" * 60)
print("3. ASSOCIATION RULES")
print("-" * 60)

rules = pd.DataFrame()

if len(frequent_itemsets) >= 2:

    with warnings.catch_warnings():
        warnings.simplefilter("ignore")

        rules = association_rules(
            frequent_itemsets,
            metric="confidence",
            min_threshold=0.10
        )

if not rules.empty:

    rules = rules[
        [
            "antecedents",
            "consequents",
            "support",
            "confidence",
            "lift"
        ]
    ]

    rules = rules.sort_values(
        by="lift",
        ascending=False
    ).reset_index(drop=True)

    print("Number of association rules:", len(rules))

    display(rules.head(15))

else:
    print("No association rules found.")

# ------------------------------------------------------------
# 9. FINAL SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("TASK 4 SUMMARY")
print("=" * 60)

print("Total transactions :", basket.shape[0])
print("Total items        :", basket.shape[1])
print("Frequent itemsets  :", len(frequent_itemsets))
print("Association rules  :", len(rules))

print("\n" + "=" * 60)
print("TASK 4 COMPLETED SUCCESSFULLY")
print("=" * 60)
```

```python
# ============================================================
# TASK 5 - FOOD RECOMMENDATION SYSTEM
# ============================================================

import pandas as pd

# Import Apriori
try:
    from mlxtend.frequent_patterns import apriori, association_rules
except ImportError:
    !pip install mlxtend -q
    from mlxtend.frequent_patterns import apriori, association_rules

print("=" * 60)
print("TASK 5 - FOOD RECOMMENDATION SYSTEM")
print("=" * 60)

# ------------------------------------------------------------
# 1. Load dataset
# ------------------------------------------------------------
df = pd.read_csv(
    "food delivery analytics.csv",
    sep=None,
    engine="python"
)

print("\nDataset loaded successfully!")
print("Shape:", df.shape)

# ------------------------------------------------------------
# 2. Create transaction basket
# ------------------------------------------------------------
basket = (
    df.groupby(["OrderID", "FoodItem"])["Quantity"]
      .sum()
      .unstack(fill_value=0)
)

# Convert quantity to 0 and 1
basket = basket.map(lambda x: 1 if x > 0 else 0)

print("\n1. TRANSACTION BASKET CREATED")
print("Number of transactions:", len(basket))
print("Number of food items:", len(basket.columns))

# 2. Generate frequent itemsets
print("\n2. FREQUENT ITEMSETS")

frequent_itemsets = apriori(
    basket,
    min_support=0.01,
    use_colnames=True
)

if not frequent_itemsets.empty:
    frequent_itemsets = frequent_itemsets.sort_values(
        by="support",
        ascending=False
    )

    print("Frequent itemsets generated successfully!")
    display(frequent_itemsets.head(20))

# ------------------------------------------------------------
# 4. Generate association rules
# ------------------------------------------------------------
rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.20
)

print("\n3. ASSOCIATION RULES")
print("Number of rules:", len(rules))

if not rules.empty:
    display(
        rules[
            [
                "antecedents",
                "consequents",
                "support",
                "confidence",
                "lift"
            ]
        ]
        .sort_values(
            ["lift", "confidence"],
            ascending=False
        )
        .head(10)
    )

# ------------------------------------------------------------
# 5. Recommendation function
# ------------------------------------------------------------
def recommend_food(food_item, number_of_recommendations=5):

    # Check whether food exists
    if food_item not in basket.columns:
        print("\nFood item not found:", food_item)
        print("\nAvailable food items:")
        print(", ".join(sorted(basket.columns)))
        return []

    # Find rules containing selected food
    recommendations = rules[
        rules["antecedents"].map(
            lambda x: food_item in x
        )
    ].copy()

    if recommendations.empty:
        print("\nNo recommendations found for:", food_item)
        return []

    # Sort by lift and confidence
    recommendations = recommendations.sort_values(
        ["lift", "confidence"],
        ascending=False
    )

    result = []

    for _, row in recommendations.iterrows():

        for item in row["consequents"]:

            if item != food_item and item not in result:
                result.append(item)

            if len(result) >= number_of_recommendations:
                break

        if len(result) >= number_of_recommendations:
            break

    print("\nRECOMMENDATIONS")
    print("-" * 40)
    print("Selected food:", food_item)

    if result:
        for i, item in enumerate(result, 1):
            print(f"{i}. {item}")
    else:
        print("No recommendations found.")

    return result


# ------------------------------------------------------------
# 6. Test recommendation system
# ------------------------------------------------------------
print("\n4. TESTING RECOMMENDATION SYSTEM")

recommend_food("Pizza", 5)

print("\n" + "=" * 60)
print("TASK 5 COMPLETED SUCCESSFULLY")
print("=" * 60)
```

```python
# ============================================================
# TASK 6 - TIME SERIES FORECASTING
# ============================================================

import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.arima.model import ARIMA

print("=" * 60)
print("TASK 6 - TIME SERIES FORECASTING")
print("=" * 60)

# ------------------------------------------------------------
# 1. USE EXISTING DATAFRAME
# ------------------------------------------------------------

try:
    data = df.copy()
except NameError:
    raise ValueError(
        "Please run your dataset loading cell first so that 'df' exists."
    )

# Clean column names
data.columns = (
    data.columns.astype(str)
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print("\nColumns available:")
print(data.columns.tolist())


# ------------------------------------------------------------
# 2. FIND DATE COLUMN
# ------------------------------------------------------------

date_col = None

for col in data.columns:
    if "date" in col or "month" in col:
        date_col = col
        break

if date_col is None:
    raise ValueError(
        "Date/Month column was not found. "
        "Please check your dataset columns."
    )


# ------------------------------------------------------------
# 3. FIND SALES COLUMN
# ------------------------------------------------------------

sales_col = None

for col in data.columns:
    if col in ["sales", "sale", "revenue", "amount", "total_sales"]:
        sales_col = col
        break

if sales_col is None:
    # Try numeric columns
    numeric_cols = data.select_dtypes(
        include=np.number
    ).columns.tolist()

    if len(numeric_cols) > 0:
        sales_col = numeric_cols[-1]

if sales_col is None:
    raise ValueError(
        "Sales column was not found."
    )

print("\nDate column :", date_col)
print("Sales column:", sales_col)


# ------------------------------------------------------------
# 4. CONVERT DATA
# ------------------------------------------------------------

data[date_col] = pd.to_datetime(
    data[date_col],
    errors="coerce"
)

data[sales_col] = pd.to_numeric(
    data[sales_col],
    errors="coerce"
)

data = data.dropna(
    subset=[date_col, sales_col]
).copy()

data = data.sort_values(date_col)


# ------------------------------------------------------------
# 5. MONTHLY SALES
# ------------------------------------------------------------

monthly_sales = (
    data.set_index(date_col)[sales_col]
    .resample("MS")
    .sum()
)

monthly_sales = monthly_sales.dropna()

print("\n1. MONTHLY SALES")
print("-" * 60)
print(monthly_sales)


# ------------------------------------------------------------
# 6. TREND ANALYSIS
# ------------------------------------------------------------

print("\n2. TREND ANALYSIS")
print("-" * 60)

first_sales = monthly_sales.iloc[0]
last_sales = monthly_sales.iloc[-1]

if last_sales > first_sales:
    trend = "Increasing"
elif last_sales < first_sales:
    trend = "Decreasing"
else:
    trend = "Stable"

print("First Month Sales :", round(first_sales, 2))
print("Last Month Sales  :", round(last_sales, 2))
print("Overall Trend     :", trend)


# ------------------------------------------------------------
# 7. MONTHLY TREND GRAPH
# ------------------------------------------------------------

plt.figure(figsize=(12, 5))

plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 8. ARIMA MODEL
# ------------------------------------------------------------

print("\n3. ARIMA MODEL")
print("-" * 60)

if len(monthly_sales) < 5:
    raise ValueError(
        "At least 5 months of sales data are required."
    )

model = ARIMA(
    monthly_sales,
    order=(1, 1, 1)
)

model_fit = model.fit()

print("ARIMA(1,1,1) model fitted successfully.")


# ------------------------------------------------------------
# 9. FORECAST NEXT 3 MONTHS
# ------------------------------------------------------------

forecast = model_fit.forecast(steps=3)

future_months = pd.date_range(
    start=monthly_sales.index[-1]
    + pd.offsets.MonthBegin(1),
    periods=3,
    freq="MS"
)

forecast_df = pd.DataFrame({
    "Month": future_months,
    "Forecast_Sales": np.asarray(forecast)
})

print("\n4. NEXT 3 MONTH SALES FORECAST")
print("-" * 60)

print(
    forecast_df.to_string(
        index=False,
        formatters={
            "Forecast_Sales": "{:.2f}".format
        }
    )
)


# ------------------------------------------------------------
# 10. ACTUAL + FORECAST GRAPH
# ------------------------------------------------------------

plt.figure(figsize=(12, 6))

plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker="o",
    label="Actual Sales"
)

plt.plot(
    future_months,
    forecast_df["Forecast_Sales"],
    marker="o",
    linestyle="--",
    label="Forecast Sales"
)

plt.title("Sales Forecast - Next 3 Months")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


print("\n" + "=" * 60)
print("TASK 6 COMPLETED SUCCESSFULLY")
print("=" * 60)
```

```python
# ============================================================
# TASK 7 - FASTAPI RECOMMENDATION API
# ============================================================

import os

print("=" * 60)
print("TASK 7 - FASTAPI RECOMMENDATION API")
print("=" * 60)

# ------------------------------------------------------------
# 1. Create app.py
# ------------------------------------------------------------

app_code = r'''
import pandas as pd
from fastapi import FastAPI
from pydantic import BaseModel
from mlxtend.frequent_patterns import apriori, association_rules

# Load dataset
df = pd.read_csv(
    "food delivery analytics.csv",
    sep=None,
    engine="python"
)

# Create transaction basket
basket = (
    df.groupby(["OrderID", "FoodItem"])["Quantity"]
      .sum()
      .unstack(fill_value=0)
)

# Convert quantity to 0 and 1
basket = basket.map(lambda x: 1 if x > 0 else 0)

# Generate frequent itemsets
frequent_itemsets = apriori(
    basket,
    min_support=0.03,
    use_colnames=True
)

# Generate association rules
rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.20
)

# Create FastAPI application
app = FastAPI(
    title="Food Delivery Recommendation API",
    description="Food recommendation system using Apriori",
    version="1.0"
)


class FoodRequest(BaseModel):
    food_item: str
    number_of_recommendations: int = 5


@app.get("/")
def home():
    return {
        "message": "Food Delivery Recommendation API is running"
    }


@app.post("/recommend")
def recommend(request: FoodRequest):

    food_item = request.food_item
    number_of_recommendations = request.number_of_recommendations

    # Check food item
    if food_item not in basket.columns:
        return {
            "food_item": food_item,
            "recommendations": [],
            "message": "Food item not found"
        }

    # Find matching rules
    recommendations = rules[
        rules["antecedents"].map(
            lambda x: food_item in x
        )
    ].copy()

    if recommendations.empty:
        return {
            "food_item": food_item,
            "recommendations": []
        }

    # Sort by lift and confidence
    recommendations = recommendations.sort_values(
        ["lift", "confidence"],
        ascending=False
    )

    result = []

    for _, row in recommendations.iterrows():

        for item in row["consequents"]:

            if item != food_item and item not in result:
                result.append(item)

            if len(result) >= number_of_recommendations:
                break

        if len(result) >= number_of_recommendations:
            break

    return {
        "food_item": food_item,
        "recommendations": result
    }
'''

with open("app.py", "w", encoding="utf-8") as file:
    file.write(app_code)

# ------------------------------------------------------------
# 2. Create requirements.txt
# ------------------------------------------------------------

requirements = """pandas
mlxtend
fastapi
uvicorn
pydantic
"""

with open("requirements.txt", "w", encoding="utf-8") as file:
    file.write(requirements)

# ------------------------------------------------------------
# 3. Create Dockerfile
# ------------------------------------------------------------

dockerfile = """FROM python:3.11-slim

WORKDIR /app

COPY requirements.txt .

RUN pip install --no-cache-dir -r requirements.txt

COPY app.py .
COPY food delivery analytics.csv .

EXPOSE 8000

CMD ["uvicorn", "app:app", "--host", "0.0.0.0", "--port", "8000"]
"""

with open("Dockerfile", "w", encoding="utf-8") as file:
    file.write(dockerfile)

# ------------------------------------------------------------
# 4. Check files
# ------------------------------------------------------------

print("\nFILES CREATED SUCCESSFULLY:")
print("1. app.py")
print("2. requirements.txt")
print("3. Dockerfile")

print("\n" + "=" * 60)
print("TASK 7 COMPLETED SUCCESSFULLY")
print("=" * 60)
```
